```python
# import libraries

import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# define the path to the data files

content_path = "citeseer.content"
cites_path = "citeseer.cites"

# verify path
print(Path(content_path).exists())
print(Path(cites_path).exists())
```

```python
# inspect the content file
with open(content_path, "r", encoding="utf-8") as f:
    for _ in range(5):
        print(repr(next(f)))
```

```python
# inspect cites file
with open(cites_path, "r", encoding="utf-8") as f:
    for _ in range(5):
        print(repr(next(f)))
```

```python
# load content file
content = pd.read_csv(
    content_path,
    sep="\t",
    header=None
)

print("Shape:", content.shape)
print(content.head())
```

```python
# load cites file
cites = pd.read_csv(
    cites_path,
    sep="\t",
    header=None,
    dtype=str
)

print("Shape:", cites.shape)
print(cites.head())
```

```python
# inspect the structure
print("Number of nodes:", content.shape[0])
print("Number of columns:", content.shape[1])

print("\nFirst column:")
print(content.iloc[:5, 0])

print("\nLast column:")
print(content.iloc[:5, -1])

print("\nNumber of citation edges:", cites.shape[0])
```

```python
# separate components
node_ids = content.iloc[:, 0].astype(str).values
X = content.iloc[:, 1:-1].values.astype(np.float32)
labels = content.iloc[:, -1].values

print("Node IDs shape:", node_ids.shape)
print("Feature matrix shape:", X.shape)
print("Labels shape:", labels.shape)

print("\nClasses:")
print(np.unique(labels))
```

```python
# check class distribution
unique_labels, counts = np.unique(labels, return_counts=True)

for label, count in zip(unique_labels, counts):
    print(f"{label}: {count}")
```

```python
# inspect citation IDs
cite_sources = cites.iloc[:, 0].astype(str)
cite_targets = cites.iloc[:, 1].astype(str)

all_cite_ids = set(cite_sources) | set(cite_targets)
content_ids = set(node_ids)

missing_ids = all_cite_ids - content_ids

print("Unique citation IDs:", len(all_cite_ids))
print("Missing IDs:", len(missing_ids))
print("Examples of missing IDs:", list(missing_ids)[:10])
```

```python
# build the ID-row index mapping
node_to_index = {
    node_id: index
    for index, node_id in enumerate(node_ids)
}

print("Number of mapped nodes:", len(node_to_index))
```

```python
# filter citation edges
valid_edges = []

for source, target in zip(cite_sources, cite_targets):
    if source in node_to_index and target in node_to_index:
        valid_edges.append((source, target))

print("Original citation edges:", len(cites))
print("Valid citation edges:", len(valid_edges))

# check how many were removed
print("Removed edges:", len(cites) - len(valid_edges))
```

```python
# build undirected graph
neighbors = {
    node_id: set()
    for node_id in node_ids
}

for source, target in valid_edges:
    neighbors[source].add(target)
    neighbors[target].add(source)
```

```python
# examine neighborhood size
degrees = np.array([
    len(neighbors[node_id])
    for node_id in node_ids
])

print("Minimum neighbors:", degrees.min())
print("Maximum neighbors:", degrees.max())
print("Average neighbors:", degrees.mean())
print("Median neighbors:", np.median(degrees))
```

```python
# inspect distribution

print("Nodes with no neighbors:", np.sum(degrees == 0))
print("Nodes with 1 neighbor:", np.sum(degrees == 1))
print("Nodes with 2 neighbors:", np.sum(degrees == 2))
print("Nodes with 3 neighbors:", np.sum(degrees == 3))
print("Nodes with 4 neighbors:", np.sum(degrees == 4))
print("Nodes with 5+ neighbors:", np.sum(degrees >= 5))
```

```python
# create feature lookup
feature_lookup = {
    node_id: X[index]
    for node_id, index in node_to_index.items()
}

print("Number of feature vectors:", len(feature_lookup))
print("Feature size:", len(next(iter(feature_lookup.values()))))
```

```python
# test one node
node_id = node_ids[0]

center_features = feature_lookup[node_id]

neighbor_features = np.array([
    feature_lookup[neighbor_id]
    for neighbor_id in neighbors[node_id]
])

print("Node:", node_id)
print("Center shape:", center_features.shape)
print("Neighbor shape:", neighbor_features.shape)
```

```python
# encode class labels

label_to_index = {
    label: index
    for index, label in enumerate(np.unique(labels))
}

y = np.array([
    label_to_index[label]
    for label in labels
])

print(label_to_index)
print("Encoded labels:", y[:10])
```

```python
# stratified node splitting
from sklearn.model_selection import train_test_split

node_indices = np.arange(len(node_ids))

train_indices, test_indices = train_test_split(
    node_indices,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training nodes:", len(train_indices))
print("Test nodes:", len(test_indices))
```

```python
# verify class distribution in splits
print("Training class counts:")
print(np.bincount(y[train_indices]))

print("\nTest class counts:")
print(np.bincount(y[test_indices]))
```

```python
# save prepared arrays
np.savez(
    "citeseer_prepared.npz",
    X=X,
    y=y,
    train_indices=train_indices,
    test_indices=test_indices,
    node_ids=node_ids
)

print("Citeseer data saved.")
```
